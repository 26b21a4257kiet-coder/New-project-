```python

# ============================================================
# FACTORY PRODUCTION PLANNER (A7)
# B.Tech Engineering Mathematics Mini Project
# ============================================================

# ------------------------------------------------------------
# 1. IMPORTS
# ------------------------------------------------------------

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from IPython.display import display, Markdown


# ------------------------------------------------------------
# 2. PROJECT INPUTS
# ------------------------------------------------------------
# Mathematical assumption:
# A = machine-time requirements
# X = production quantities
# B = available machine hours
#
# Mathematical model:
#                 AX = B
#
# For an inconsistent system, a least-squares solution is used
# ONLY as a diagnostic to identify the machine with the largest
# normalized residual.


products = ["Product A", "Product B", "Product C"]

machines = ["Machine 1", "Machine 2", "Machine 3"]


# ------------------------------------------------------------
# BENCHMARK DATASET
# ------------------------------------------------------------
#
#              Product A   Product B   Product C
# Machine 1        2            1           1
# Machine 2        1            2           1
# Machine 3        1            1           2
#
# Available machine hours:
# Machine 1 = 80
# Machine 2 = 100
# Machine 3 = 70
#
# Expected solution:
# Product A = 20
# Product B = 30
# Product C = 10
# ------------------------------------------------------------

machine_times = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
], dtype=float)

available_hours = np.array([
    80,
    100,
    70
], dtype=float)


# ------------------------------------------------------------
# 3. CONSTRUCT MATHEMATICAL MODEL
# ------------------------------------------------------------

A = sp.Matrix(machine_times.astype(int))
B = sp.Matrix(available_hours.astype(int))

X_symbols = sp.Matrix([
    sp.Symbol("x1"),
    sp.Symbol("x2"),
    sp.Symbol("x3")
])

print("=" * 70)
print("             FACTORY PRODUCTION PLANNER (A7)")
print("=" * 70)

print("\nPROJECT INPUTS")
print("-" * 70)

print("\nProducts:")
for p in products:
    print("  ", p)

print("\nMachines:")
for m in machines:
    print("  ", m)

print("\nMachine-Time Matrix A:")
display(A)

print("\nAvailable Hours Vector B:")
display(B)

print("\nMathematical Model:")
print("                 AX = B")
display(sp.Eq(A * X_symbols, B))


# ------------------------------------------------------------
# 4. DISPLAY LINEAR EQUATIONS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FACTORY LINEAR EQUATIONS")
print("=" * 70)

for i in range(A.rows):

    equation = ""

    for j in range(A.cols):

        coefficient = A[i, j]

        if coefficient != 0:

            if equation != "" and coefficient > 0:
                equation += " + "

            equation += f"{coefficient}{X_symbols[j]}"

    equation += f" = {B[i]}"

    print(f"{machines[i]}: {equation}")


# ------------------------------------------------------------
# 5. RANK ANALYSIS
# ------------------------------------------------------------

augmented_matrix = A.row_join(B)

rank_A = A.rank()
rank_augmented = augmented_matrix.rank()

number_of_unknowns = A.cols

print("\n" + "=" * 70)
print("CONSISTENCY AND RANK ANALYSIS")
print("=" * 70)

print("\nMatrix A:")
display(A)

print("\nAugmented Matrix [A | B]:")
display(augmented_matrix)

print(f"\nRank(A)              = {rank_A}")
print(f"Rank([A | B])        = {rank_augmented}")
print(f"Number of unknowns   = {number_of_unknowns}")


# ------------------------------------------------------------
# 6. CLASSIFY THE SYSTEM
# ------------------------------------------------------------

if rank_A != rank_augmented:

    system_status = "INCONSISTENT"

elif rank_A == number_of_unknowns:

    system_status = "CONSISTENT - UNIQUE SOLUTION"

else:

    system_status = "CONSISTENT - INFINITELY MANY SOLUTIONS"


print(f"\nSystem Status: {system_status}")


# ------------------------------------------------------------
# 7. SOLVE THE SYSTEM
# ------------------------------------------------------------

solution = None
approx_solution = None

print("\n" + "=" * 70)
print("MATHEMATICAL SOLUTION")
print("=" * 70)


if rank_A != rank_augmented:

    print("\nThe system is INCONSISTENT.")
    print("No exact production quantities satisfy all equations.")

    # Least-squares solution is used only for diagnostic analysis
    A_numpy = np.array(A).astype(float)
    B_numpy = np.array(B).astype(float).reshape(-1)

    approx_solution, residuals, rank, singular_values = np.linalg.lstsq(
        A_numpy,
        B_numpy,
        rcond=None
    )

    print("\nLeast-Squares Diagnostic Solution:")

    for product, quantity in zip(products, approx_solution):

        print(f"  {product}: {quantity:.4f}")


elif rank_A == number_of_unknowns:

    # Unique exact solution
    solution_vector = A.inv() * B

    solution = np.array(
        solution_vector,
        dtype=float
    ).reshape(-1)

    print("\nUnique Exact Solution Found:")

    for product, quantity in zip(products, solution):

        print(f"  {product}: {quantity:.4f}")


else:

    print("\nThe system has infinitely many solutions.")
    print("A unique production plan cannot be determined.")


# ------------------------------------------------------------
# 8. VERIFY EXACT SOLUTION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SOLUTION VERIFICATION")
print("=" * 70)

if solution is not None:

    A_numpy = np.array(A).astype(float)
    B_numpy = np.array(B).astype(float).reshape(-1)

    calculated_hours = A_numpy @ solution

    print(
        f"\n{'Machine':<15}"
        f"{'Available':<15}"
        f"{'Calculated':<15}"
        f"{'Difference':<15}"
    )

    print("-" * 60)

    for i in range(len(machines)):

        difference = calculated_hours[i] - B_numpy[i]

        print(
            f"{machines[i]:<15}"
            f"{B_numpy[i]:<15.2f}"
            f"{calculated_hours[i]:<15.2f}"
            f"{difference:<15.6f}"
        )

    if np.allclose(calculated_hours, B_numpy):

        print("\n✓ Verification Successful!")
        print("The production quantities satisfy AX = B.")

    else:

        print("\n✗ Verification Failed.")

else:

    print("\nNo exact solution is available for verification.")


# ------------------------------------------------------------
# 9. PRODUCTION PLAN
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FACTORY PRODUCTION PLAN")
print("=" * 70)

if solution is not None:

    print(
        f"\n{'Product':<25}"
        f"{'Production Quantity':>20}"
    )

    print("-" * 45)

    for product, quantity in zip(products, solution):

        print(
            f"{product:<25}"
            f"{quantity:>20.2f}"
        )

else:

    print("\nNo unique exact production plan is available.")


# ------------------------------------------------------------
# 10. PRACTICAL INTERPRETATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PRACTICAL INTERPRETATION")
print("=" * 70)

if solution is not None:

    print("\nThe factory can produce:")

    for product, quantity in zip(products, solution):

        print(
            f"  • {product}: {quantity:.2f} units"
        )

    print(
        "\nThe calculated production quantities satisfy "
        "all three machine-hour equations."
    )

elif approx_solution is not None:

    print(
        "\nThe system is inconsistent."
    )

    print(
        "Therefore, no exact production plan can satisfy "
        "all machine-hour requirements simultaneously."
    )


# ------------------------------------------------------------
# 11. BOTTLENECK ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BOTTLENECK ANALYSIS")
print("=" * 70)

bottleneck_index = None
normalized_residuals = None

if approx_solution is not None:

    A_numpy = np.array(A).astype(float)
    B_numpy = np.array(B).astype(float).reshape(-1)

    calculated_hours = A_numpy @ approx_solution

    # Residual:
    # calculated machine hours - available machine hours
    residuals = calculated_hours - B_numpy

    # Normalize residual by available machine hours
    normalized_residuals = np.where(
        B_numpy != 0,
        np.abs(residuals) / np.abs(B_numpy),
        np.abs(residuals)
    )

    bottleneck_index = np.argmax(normalized_residuals)

    print(
        f"\n{'Machine':<15}"
        f"{'Calculated':<15}"
        f"{'Available':<15}"
        f"{'Residual':<15}"
        f"{'Normalized':<15}"
    )

    print("-" * 75)

    for i in range(len(machines)):

        print(
            f"{machines[i]:<15}"
            f"{calculated_hours[i]:<15.2f}"
            f"{B_numpy[i]:<15.2f}"
            f"{residuals[i]:<15.2f}"
            f"{normalized_residuals[i]:<15.4f}"
        )

    print("\nBottleneck Diagnostic:")
    print(
        f"→ {machines[bottleneck_index]} "
        f"has the largest normalized residual."
    )

    print(
        "\nTherefore, under the stated residual-based diagnostic "
        "assumption, this machine is identified as the bottleneck."
    )

else:

    print(
        "\nThe exact system is consistent."
    )

    print(
        "There is no inconsistency-based bottleneck to diagnose."
    )


# ------------------------------------------------------------
# 12. MACHINE UTILIZATION
# ------------------------------------------------------------

if solution is not None:

    A_numpy = np.array(A).astype(float)
    B_numpy = np.array(B).astype(float).reshape(-1)

    used_hours = A_numpy @ solution

    utilization = (used_hours / B_numpy) * 100

    print("\n" + "=" * 70)
    print("MACHINE UTILIZATION")
    print("=" * 70)

    for machine, percentage in zip(machines, utilization):

        print(
            f"{machine}: {percentage:.2f}%"
        )


# ------------------------------------------------------------
# 13. CHART 1 - MACHINE UTILIZATION
# ------------------------------------------------------------

if solution is not None:

    plt.figure(figsize=(9, 5))

    bars = plt.bar(
        machines,
        utilization
    )

    plt.axhline(
        100,
        linestyle="--",
        linewidth=1.5,
        label="100% Capacity"
    )

    plt.title(
        "Factory Machine Utilization"
    )

    plt.xlabel(
        "Machines"
    )

    plt.ylabel(
        "Utilization (%)"
    )

    plt.ylim(
        0,
        max(120, max(utilization) + 10)
    )

    plt.legend()

    plt.grid(
        axis="y",
        alpha=0.3
    )

    for bar, value in zip(bars, utilization):

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 2,
            f"{value:.1f}%",
            ha="center"
        )

    plt.show()


# ------------------------------------------------------------
# 14. CHART 2 - PRODUCTION QUANTITIES
# ------------------------------------------------------------

if solution is not None:

    plt.figure(figsize=(9, 5))

    bars = plt.bar(
        products,
        solution
    )

    plt.title(
        "Factory Production Plan"
    )

    plt.xlabel(
        "Products"
    )

    plt.ylabel(
        "Production Quantity"
    )

    for bar, value in zip(bars, solution):

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f"{value:.2f}",
            ha="center",
            va="bottom"
        )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.show()


# ------------------------------------------------------------
# 15. CHART 3 - AVAILABLE VS USED HOURS
# ------------------------------------------------------------

if solution is not None:

    A_numpy = np.array(A).astype(float)
    B_numpy = np.array(B).astype(float).reshape(-1)

    used_hours = A_numpy @ solution

    x = np.arange(len(machines))

    width = 0.35

    plt.figure(figsize=(10, 5))

    plt.bar(
        x - width / 2,
        B_numpy,
        width,
        label="Available Hours"
    )

    plt.bar(
        x + width / 2,
        used_hours,
        width,
        label="Used Hours"
    )

    plt.xticks(
        x,
        machines
    )

    plt.xlabel(
        "Machines"
    )

    plt.ylabel(
        "Hours"
    )

    plt.title(
        "Machine Capacity vs Production Requirement"
    )

    plt.legend()

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.show()


# ------------------------------------------------------------
# 16. CHART 4 - BOTTLENECK RESIDUALS
# ------------------------------------------------------------

if approx_solution is not None:

    plt.figure(figsize=(9, 5))

    plt.bar(
        machines,
        normalized_residuals * 100
    )

    plt.title(
        "Normalized Machine Residuals"
    )

    plt.xlabel(
        "Machines"
    )

    plt.ylabel(
        "Residual (%)"
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.show()


# ------------------------------------------------------------
# 17. FINAL PROJECT SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("                 PROJECT SUMMARY")
print("=" * 70)

print(
    "\nMathematical Model:"
)

print(
    "                    AX = B"
)

print(
    "\nRank(A)       =",
    rank_A
)

print(
    "Rank([A|B])   =",
    rank_augmented
)

print(
    "\nSystem Status:"
)

print(
    system_status
)


if solution is not None:

    print(
        "\nFinal Production Quantities:"
    )

    for product, quantity in zip(products, solution):

        print(
            f"  {product}: {quantity:.2f} units"
        )

    print(
        "\nFinal Conclusion:"
    )

    print(
        "The system is consistent and has a unique production plan."
    )

    print(
        "The calculated quantities satisfy all machine-hour equations."
    )

elif approx_solution is not None:

    print(
        "\nFinal Conclusion:"
    )

    print(
        "The system is inconsistent."
    )

    print(
        "No exact production plan satisfies all machine equations."
    )

    print(
        f"{machines[bottleneck_index]} is identified as the "
        "bottleneck using the normalized-residual diagnostic."
    )

else:

    print(
        "\nFinal Conclusion:"
    )

    print(
        "The system does not provide a unique production plan."
    )


print("\n" + "=" * 70)
print("             END OF FACTORY PLANNER")
print("=" * 70)


# ============================================================
# OPTIONAL INCONSISTENT TEST
# ============================================================
#
# To test the bottleneck feature, replace the input section
# above with the following data and run the complete code again:
#
# machine_times = np.array([
#     [2, 1, 1],
#     [1, 2, 1],
#     [3, 3, 2]
# ], dtype=float)
#
# available_hours = np.array([
#     80,
#     100,
#     190
# ], dtype=float)
#
# Here:
#
# Row 3 = Row 1 + Row 2
#
# But:
#
# 190 != 80 + 100
#
# Therefore:
#
# rank(A) != rank([A|B])
#
# and the system becomes INCONSISTENT.
# ============================================================
```
