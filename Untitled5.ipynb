```python

# ============================================================
# FACTORY PRODUCTION PLANNER (A7)
# Team Customization + Rank/Consistency + Sensitivity Analysis
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# -------------------- TEAM HEADER --------------------
TEAM_NAME = "TEAM A7"
COLLEGE_NAME = "KIET Engineering College"
PROJECT_TITLE = "Factory Production Planner (A7)"
BENCHMARK = "Custom Benchmark: Manufacturing Plant"

print("=" * 65)
print(PROJECT_TITLE)
print("=" * 65)
print(f"Team        : {TEAM_NAME}")
print(f"College     : {COLLEGE_NAME}")
print(f"Benchmark   : {BENCHMARK}")
print("=" * 65)

# -------------------- CUSTOM DATASET --------------------
# Products: Product A, Product B, Product C
# Machines: Machine 1, Machine 2, Machine 3
#
# A[i,j] = hours required by Machine i for Product j
# B[i]   = available machine hours

A = np.array([
    [2, 1, 3],   # Machine 1
    [1, 2, 2],   # Machine 2
    [3, 2, 1]    # Machine 3
], dtype=float)

B = np.array([120, 100, 110], dtype=float)

products = ["Product A", "Product B", "Product C"]
machines = ["Machine 1", "Machine 2", "Machine 3"]

# -------------------- MATHEMATICAL MODEL --------------------
# AX = B
#
# 2x + y + 3z = 120
# x + 2y + 2z = 100
# 3x + 2y + z = 110

x, y, z = sp.symbols("x y z")
X = sp.Matrix([x, y, z])

A_sp = sp.Matrix(A.astype(int))
B_sp = sp.Matrix(B.astype(int))

print("\nSYSTEM OF LINEAR EQUATIONS")
print("A X = B")
sp.pprint(A_sp)
print("X =")
sp.pprint(X)
print("B =")
sp.pprint(B_sp)

# -------------------- CONSISTENCY + RANK ANALYSIS --------------------
rank_A = A_sp.rank()
augmented = A_sp.row_join(B_sp)
rank_aug = augmented.rank()

print("\nCONSISTENCY AND RANK ANALYSIS")
print(f"Rank(A)        = {rank_A}")
print(f"Rank([A|B])    = {rank_aug}")

if rank_A == rank_aug:
    print("System is CONSISTENT.")
else:
    print("System is INCONSISTENT.")

# -------------------- SOLUTION --------------------
if rank_A == rank_aug == 3:
    solution = A_sp.inv() * B_sp

    print("\nPRODUCTION QUANTITIES")
    for i, p in enumerate(products):
        print(f"{p}: {float(solution[i]):.2f} units")

    # Verify AX = B
    verification = A_sp * solution

    print("\nVERIFICATION (AX = B)")
    sp.pprint(verification)

    # -------------------- MACHINE UTILIZATION --------------------
    production = np.array([float(solution[i]) for i in range(3)])

    used_hours = A @ production
    utilization = (used_hours / B) * 100

    print("\nMACHINE UTILIZATION")
    for i in range(3):
        print(
            f"{machines[i]}: "
            f"{used_hours[i]:.2f}/{B[i]:.2f} hours "
            f"({utilization[i]:.2f}%)"
        )

    # Practical bottleneck:
    # For an exact AX=B model, all machine capacities are fully used.
    # For practical comparison, the machine with the highest
    # utilization percentage is identified as the bottleneck.
    bottleneck_index = np.argmax(utilization)

    print(
        f"\nPRACTICAL INTERPRETATION: "
        f"{machines[bottleneck_index]} is the bottleneck "
        f"based on highest utilization."
    )

    # -------------------- EXTRA ANALYSIS --------------------
    # Sensitivity analysis:
    # Increase Machine 1 available hours and observe Product A output.

    extra_hours = np.arange(0, 61, 10)
    product_A_values = []

    for extra in extra_hours:
        B_test = B.copy()
        B_test[0] += extra

        try:
            test_solution = np.linalg.solve(A, B_test)
            product_A_values.append(test_solution[0])
        except np.linalg.LinAlgError:
            product_A_values.append(np.nan)

    # -------------------- VISUALIZATION --------------------
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Machine utilization chart
    axes[0].bar(machines, utilization)
    axes[0].set_title("Machine Utilization")
    axes[0].set_xlabel("Machines")
    axes[0].set_ylabel("Utilization (%)")
    axes[0].set_ylim(0, max(110, max(utilization) + 10))
    axes[0].legend(["Utilization"], loc="best")
    axes[0].grid(axis="y", alpha=0.3)

    # Sensitivity plot
    axes[1].plot(
        extra_hours,
        product_A_values,
        marker="o",
        label="Product A output"
    )
    axes[1].set_title(
        "Sensitivity: Machine 1 Capacity vs Product A Output"
    )
    axes[1].set_xlabel("Additional Machine 1 Hours")
    axes[1].set_ylabel("Product A Units")
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)

    fig.suptitle(
        f"{PROJECT_TITLE} | {TEAM_NAME}",
        fontsize=15,
        fontweight="bold"
    )

    plt.tight_layout()
    plt.show()

else:
    print("\nNo unique production quantities exist.")
    print(
        "Use the rank comparison to determine whether the system "
        "is consistent or inconsistent."
    )
```
